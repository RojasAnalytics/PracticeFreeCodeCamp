```python
import pandas as pd
import numpy as np  

print('Pandas Version Nro:\n', {pd.__version__})
print('NumPy Version Nro:\n', {np.__version__})
```

```python
df_videogames_repaso = pd.read_csv('vgsales.csv')

df_videogames_repaso.head()
```

```python
df_videogames_repaso.info()
```

Se observan en una primera impresion valores nulos en las columnas de 'Year' y 'Publisher', lo que puede afectar el analisis de los datos.

```python
df_videogames_repaso.describe()
```

Se obverva que la desviacion estandar de las ventas globales es muy alta, lo que indica que hay una gran variabilidad en las ventas de los videojuegos. Esto sugiere que algunos videojuegos se venden mucho mejor que otros, lo que puede deberse a factores como la popularidad del juego, la calidad del marketing, o la competencia en el mercado, tambien puede deberse a numeroras ventas por lanzamiento que lo que tambien podria afectar las ventas globales.

```python
df_videogames_repaso.isna().sum()
```

Se confirman 271 datos nulos para la columna 'Year' y 58 para la columna 'Publisher', estos seran reemplazos con el el año promedio para la columna 'Year' y 'Unknown' ya que a pesar de representar un valor muy porcentaje muy bajo en la base de datos, se asume que dicha informacion posee influencia en las ventas globales.

```python
mediana_year = df_videogames_repaso['Year'].median()

print('Calculo de la mediana de la columna Year:\n',{mediana_year})
```

```python
df_videogames_repaso['Year'] = df_videogames_repaso['Year'].fillna(mediana_year)
```

```python
df_videogames_repaso['Publisher'] = df_videogames_repaso['Publisher']. fillna('Unknown')
```

```python
df_videogames_repaso.isna().sum()
```

```python
df_videogames_repaso.info() 
```

```python
df_videogames_repaso['Year'] = df_videogames_repaso['Year'].astype(int)

df_videogames_repaso.info()
```

```python
df_videogames_repaso['Sales_category'] = df_videogames_repaso['Global_Sales'].apply(lambda sales: 'Existo de ventas' if sales > 1
                                                                                    else 'Ventas estandar')
df_videogames_repaso.head()
```

```python
print('Cantidad de filas duplicadas en el DataFrame:\n', {df_videogames_repaso.duplicated().sum()})
```

```python
print('Cantidad de filas duplicadas en las columnas Name y Platform:\n', {df_videogames_repaso.duplicated(subset=['Name','Platform']).sum()})
```

Se confirman 5 duplicados en la base de datos, estos seran eliminados para evitar sesgos de la informacion.

```python
df_videogames_repaso = df_videogames_repaso.drop_duplicates(subset=['Name','Platform'])

df_videogames_repaso.info()
```

```python
hits_de_nintendo = df_videogames_repaso 

hits_de_nintendo = hits_de_nintendo[(hits_de_nintendo['Publisher'] == 'Nintendo') & (hits_de_nintendo['Global_Sales'] > 1)]

hits_de_nintendo = hits_de_nintendo.sort_values(by='Global_Sales', ascending=False)

hits_de_nintendo.head(10)
```

```python
hits_de_nintendo.to_csv('hits_de_nintendo.csv', index=False)
```
